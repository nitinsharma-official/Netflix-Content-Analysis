# ***Numpy Operation's***

```python
# Convert release year into NumPy array
years = df["release_year"].to_numpy()
print(years)
```

```python
# Maximum release year
print(np.max(years))
```

```python
# Minimum release year
print(np.min(years))
```

```python
# Average release year
print(np.mean(years))
```

```python
# Median release year
print(np.median(years))
```

```python
# Count releases after 2020
print(np.sum(years > 2020))
```

```python
# Create release period using NumPy
period = np.where(years >= 2020, "Recent", "Before 2020")
print(period)
```

```python
# Add new column using NumPy result
df["release_period"] = period
print(df[["title", "release_year", "release_period"]].head())
```

```python
# Convert date_added into datetime
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)
print(df["date_added"].head())
```

```python
# Extract year when content was added to Netflix
df["date_added_year"] = df["date_added"].dt.year
print(df[["title", "date_added", "date_added_year"]].head())
```

```python
# Extract numeric duration
df["duration_value"] = pd.to_numeric(
    df["duration"].str.extract(r"(\d+)")[0],
    errors="coerce"
)
print(df[["title", "type", "duration", "duration_value"]].head())
```
